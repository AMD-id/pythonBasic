Tipe data (data type) adalah klasifikasi data yang berfungsi memberi tahu compiler atau interpreter mengenai cara data tersebut akan digunakan. Tipe data menentukan jenis operasi yang dapat dilakukan pada data, rentang nilai yang dapat disimpan, serta besarnya alokasi memori yang dibutuhkan oleh data tersebut.

```python
## integer (int)
a=10
## float (float)
score=95.5
## string (str)
name="Alice"
## boolean (bool)
status=True


## print the variables
print("a:", a, type(a))
print("score:", score, type(score))
print("name:", name, type(name))
print("status:", status, type(status))
```

```python
## list
numbers=[1, 2, 3, 4, 5]
print("numbers:", numbers, type(numbers))
```

```python
## tuple
coordinates=(10.0, 20.0)
print("coordinates:", coordinates, type(coordinates))
```

```python
## set
unique_numbers={1, 2, 3, 4, 5}
print("unique_numbers:", unique_numbers, type(unique_numbers))
```

```python
## dictionary
person={"name": "Alice", "age": 30, "city": "New York"}
print("person:", person, type(person))
```
