variabel merupakan elemen fundamental dalam promgram, digunakan untuk menyimpan data yang dapat direferensikan dan di manipulasi. variabel dibuat ketika kita memasukan value ke dalamnya. tidak perlu exlicit declaration.

```python
## declaring and assigning variables

a=100
name="am"
status=True
score=95.5

## print the variables
print("a:", a)
print("name:", name)
print("status:", status)
print("score:", score)
```

```python
## naming conventions
## Variables names should be descriptive
## must start with a letter or underscore
## case-sensitive

## Variables naming examples
first_name="John"
last_name="Doe"
```

```python
name="Alice"
Name="Bob"
print(name==Name)
```

```python
a=100
name="am"
status=True
score=95.5

## print the variables
print("a:", a, type(a))
print("name:", name, type(name))
print("status:", status, type(status))
print("score:", score, type(score))
```

```python
## type conversion
age="30"
print("age:", age, type(age))
age=int(age)
print("age:", age, type(age))
```

```python
## dynamic typing
x=10
print("x:", x, type(x))
x="Now I'm a string"
print("x:", x, type(x))
```

```python
## input

age=input("Enter your age: ")
print("your age is:", age, type(age))
```
