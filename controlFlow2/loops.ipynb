```python
## for loop

for i in range(5):
    print("i:", i)
```

```python
## dalam range ada mulai, akhir, dan step

for i in range(1, 10, 2):
    print("i:", i)
```

```python
## bisa juga negatif step untuk menghitung mundur

for i in range(10, 0, -1):
    print("i:", i)
```

```python
## string juga bisa diiterasi
for char in "Hello":
    print("char:", char)
```

```python
## while loop
## while loop akan terus dieksekusi selama kondisi yang diberikan bernilai True
count = 0
while count < 5:
    print("count:", count)
    count += 1
```

```python
## Loop control statements
## break statement
for i in range(10):
    if i == 5:
        break  # keluar dari loop ketika i sama dengan 5
    print("i:", i)
```

```python
## continue statement
for i in range(10):
    if i % 2 == 0:
        continue  # lewati iterasi ini jika i adalah angka genap
    print("i:", i)
```

```python
## pass statement
for i in range(5):
    if i == 2:
        pass  # tidak melakukan apa-apa, hanya sebagai placeholder
    print("i:", i)
```

```python
## nested loops
## a loop inside another loop
for i in range(3):
    for j in range(2):
        print(f"i: {i}, j: {j}")
```

```python
sum = 0
for i in range(1, 101):
    if i < 2:
        continue

    is_prime = True
    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            is_prime = False
            break

    if is_prime:
        sum += i

    print(f"iterasi: {i}, total : {sum}")

print(f"Jumlah bilangan prima 1-100: {sum}")
```
