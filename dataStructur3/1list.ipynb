```python
## accessing list elements
## list item itu berutut dari 0,1,2,3 dst
names=['Alice','Bob','Charlie']
print(names[0])  # Output: Alice
print(names[1])  # Output: Bob

print("-----")
## bisa juga index negatif untuk mengakses dari belakang
print(names[-1])  # Output: Charlie
```

```python
## list index

fruits=['apple','banana','cherry','date','elderberry','fig','grape','honeydew','kiwi','lemon']
print(fruits[0])  # Output: apple
print(fruits[3:])  # Output: ['date', 'elderberry', 'fig', 'grape', 'honeydew', 'kiwi', 'lemon']
print(fruits[:5])  # Output: ['apple', 'banana', 'cherry', 'date', 'elderberry']
print(fruits[2:5])  # Output: ['cherry', 'date', 'elderberry']
print(fruits[::2])  # Output: ['apple', 'cherry', 'elderberry', 'grape', 'kiwi']
```

```python
## list bisa diubah
## kita bisa menambahkan, menghapus, atau memodifikasi elemen dalam list
names=['Alice','Bob','Charlie']
names.append('David')  # Menambahkan elemen ke akhir list
print(names)  # Output: ['Alice', 'Bob', 'Charlie', 'David']
names[1] = 'Bobby'  # Memodifikasi elemen pada index 1
print(names)  # Output: ['Alice', 'Bobby', 'Charlie', 'David']
names.remove('Alice')  # Menghapus elemen 'Alice' dari list
print(names)  # Output: ['Bobby', 'Charlie', 'David']
```

```python
## list method 

fruits=['apple','banana','cherry','date']    
fruits.append('elderberry')  # Menambahkan elemen ke akhir list
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'date', 'elderberry']
```

```python
## insert
fruits.insert(1, 'blueberry')  # Menyisipkan 'blueberry' pada index 1
print(fruits)  # Output: ['apple', 'blueberry', 'banana', 'cherry', 'date', 'elderberry']
```

```python
fruits.remove('banana')  # Menghapus 'banana' dari list
print(fruits)  # Output: ['apple', 'blueberry', 'cherry', 'date', 'elderberry']
```

```python
##  pop

poped_fruit = fruits.pop(2)  # Menghapus dan mengembalikan elemen pada index 2
print(poped_fruit)  # Output: 'cherry'
```

```python
## index list
index_date = fruits.index('date')  # Mendapatkan index dari 'date'
print(index_date)  # Output: 3
```

```python
fruits.sort()  # Mengurutkan list secara alfabetis
print(fruits)  # Output: ['apple', 'blueberry', 'date', '
```

```python
fruits.reverse()  # Membalik urutan list
print(fruits)  # Output: ['elderberry', 'date', 'blueberry',
```

```python
## iterating over list
numbers=[1, 2, 3, 4, 5]
for num in numbers:
    print(num)
```

```python
## iterating with index
for i, j in enumerate(fruits):
    print(f"Index: {i}, Value: {j}")
```

```python
## list comprehension
lst=[]
for i in range(10):
    lst.append(i**2)

print(lst)
```
