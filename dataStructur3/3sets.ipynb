Sets digunakan untu menyimpan data collection yang mana data didalamnya unique. didalamnya tidak beraturan, artinya tidak menganut order, didalmnya tidak boleh diplivate. 

salah satu kegunaannya adalh membersip test, eleminating duplicate, bisa kta gunakan untuk opersi mathematika seperti union dll.

```python
## set
## sangat mirip dengan dictionary cuman bukan tipe key value
my_set={1,2,3,4,5}
print(my_set) # Output: {1, 2, 3, 4, 5}
print(type(my_set)) # Output: <class 'set'>
## set tidak memiliki urutan dan tidak bisa diakses dengan index
## set hanya menyimpan nilai unik
```

```python
# bisa juga dibuat dengan fungsi set()
my_set2=set([1,2,3,4,5])
print(my_set2) # Output: {1, 2, 3, 4, 5}
print(type(my_set2)) # Output: <class 'set'>
```

```python
## set hanya menyimpan nilai unik
my_set3={1,2,2,3,4,4,5}
print(my_set3) # Output: {1, 2, 3, 4, 5}
```

```python
## cara mengambil nilai dari set
for item in my_set:
    print(item)
    
```

```python
## add and remove item
my_set.add(6)
print(my_set) # Output: {1, 2, 3, 4, 5, 6}
my_set.remove(3)
print(my_set) # Output: {1, 2, 4, 5, 6}
```

```python
## method discard() untuk menghapus item tanpa error jika item tidak ada
my_set.discard(10) # tidak ada error meskipun item 10 tidak ada
print(my_set) # Output: {1, 2, 4, 5, 6}
```

```python
## pop untuk menghapus item secara acak
my_set.pop()
print(my_set) # Output: {2, 4, 5, 6}
my_set.pop()
print(my_set) # Output: {4, 5, 6}
```

```python
## clear untuk menghapus semua item
my_set.clear()
print(my_set) # Output: set()
```

```python
## cara mencaritahu apakah nilai ada di set
print(4 in my_set) # Output: False
print(4 in my_set3) # Output: True
```

```python
## mathematical set operations
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

## union
print(A | B) # Output: {1, 2, 3, 4, 5, 6, 7, 8}
# bisajuga
union_set = A.union(B)
print(union_set) # Output: {1, 2, 3, 4, 5, 6, 7, 8}

## intersection
print(A & B) # Output: {4, 5}
# bisajuga
intersection_set = A.intersection(B)
print(intersection_set) # Output: {4, 5}

#intersection update
A.intersection_update(B)
print(A) # Output: {4, 5}

## difference
print(A - B) # Output: {1, 2, 3}
# bisajuga
difference_set = A.difference(B)
print(difference_set) # Output: {1, 2, 3}
```

```python
## simetric difference
print(A ^ B) # Output: {1, 2, 3, 6, 7, 8}
# bisajuga
symmetric_difference_set = A.symmetric_difference(B)    
print(symmetric_difference_set) # Output: {1, 2, 3, 6, 7, 8}
```

```python
## subset dan superset
C = {4, 5}
print(C.issubset(A)) # Output: True
print(A.issuperset(C)) # Output: True
```
