Tuples are ordered collections of items that are immutable. They are similar to lists, but their immutability makes them different.

```python
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

```python
num=[1,2,3,4,5]
num_tuple=tuple(num)
print(num)
print(type(num))
print(num_tuple)
print(type(num_tuple))
```

```python
## begitujuga sebaliknya list ke tuple
tpl=(1,2,3,4,5)
lst=list(tpl)
print(tpl)
print(type(tpl))
print(lst)
print(type(lst))
```

```python
## tuple bisa mixed data type
mixed_tuple=(1, "hello", 3.14, [1,2,3], (4,5))
print(mixed_tuple)
print(type(mixed_tuple))
```

```python
## accesing tuple elements
my_tuple=(10, 20, 30, 40, 50)
print(my_tuple[0])  # Output: 10
```

```python
# indexing tuple
print(my_tuple[1])  # Output: 20
print(my_tuple[-1]) # Output: 50
print(my_tuple[1:4]) # Output: (20, 30, 40)
```

```python
## tuple operations
tuple1=(1,2,3)
tuple2=(4,5,6)
# concatenation
tuple3=tuple1+tuple2
print(tuple3) # Output: (1, 2, 3, 4, 5, 6)
```

```python
mixed_tuple=(1, "hello", 3.14, [1,2,3], (4,5))
mixed_tuple * 3
```

```python
## immutable nature of tuple
my_tuple=(10, 20, 30)
# my_tuple[0]=100 # This will raise a TypeError
```

```python
## tuple methods
my_tuple=(1,2,3,4,5)
print(my_tuple.count(2)) # Output: 1
print(my_tuple.index(3)) # Output: 2
```

```python
## packing and unpacking tuple
## jika kita menulis variable dengan nilai yang dipisahkan koma, maka itu akan otomatis menjadi tuple

packed_tuple=1, "hello", 3.14
print(packed_tuple)
print(type(packed_tuple))
```

```python
## unpacking tuple
a, b, c=packed_tuple
print(a) # Output: 1
print(b) # Output: "hello"
print(c) # Output: 3.14
```

```python
## upacking with *
numbers=(1,2,3,4,5)
a,*b,c=numbers
print(a) # Output: 1
print(b) # Output: [2, 3, 4]
print(c) # Output: 5
## a, *b, c ini bisa diganti dengan berbagai nama variable, yang penting ada * di variable tengah untuk menampung sisa elemen tuple
```

```python
## nested list
nested_tuple=(1, (2,3), [4,5], (6,7,8))
print(nested_tuple)
print(nested_tuple[2]) # Output: [4, 5]
print(nested_tuple[2][1]) # Output: 5
```

```python
## nested tuple
nested_tuple=(1, (2,3), (4,5), (6,7,8))
print(nested_tuple)
print(nested_tuple[1]) # Output: (2, 3)
print(nested_tuple[1][0]) # Output: 2
```

```python
## iterating over nested tuple
nested_tuple=(1, (2,3), (4,5), (6,7,8))
for item in nested_tuple:
    if isinstance(item, (tuple, list)):
        for sub_item in item:
            print(sub_item)
    else:
        print(item)
```
