Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
## creating dictionary
empty_dict={}
print(type(empty_dict)) # Output: <class 'dict'>
```

```python
student={"name":"Alice", "age":25, "city":"New York"}
print(student) # Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
print(type(student)) # Output: <class 'dict'>
```

```python
## accessing dictionary element
student={"name":"Alice", "age":25, "city":"New York"}
print(student)
```

```python
print(student["name"])
print(student["age"])

## accessing useing get() method
student={"name":"Alice", "age":25, "city":"New York"}
print(student.get("name"))
print(student.get("score"))
print(student.get("last_name", "Not Found")) # Output: Not Found
```

```python
## dict are mutable, kita bisa menambahkan, mengubah, dan menghapus elemen dalam dictionary
student={"name":"Alice", "age":25, "city":"New York"}
student["score"]=90
print(student) # Output: {'name': 'Alice', 'age': 25, 'city': 'New York', 'score': 90}
## update value
student["age"]=26
print(student) # Output: {'name': 'Alice', 'age': 26, 'city': 'New York', 'score': 90}
```

```python
## delete key and value pair
del student["city"]
print(student) # Output: {'name': 'Alice', 'age': 26, 'score': 90}
```

```python
## dict methods
key=student.keys() # get all key
print(key)
value=student.values() # get all value
print(value)
items=student.items() # get all key value pair as tuple
print(items)
```

```python
## jika kita melakukan copy langsng maka yang asli akan ikut berubah
student_copy=student
student_copy["name"]="Bob"
print(student) # Output: {'name': 'Bob', 'age': 26, 'score': 90}
print(student_copy) # Output: {'name': 'Bob', 'age': 26, 'score': 90}
```

```python
## karenanya kia gunakan method copy()
student_copy=student.copy()
student_copy["name"]="Charlie"
print(student) # Output: {'name': 'Bob', 'age': 26, 'score': 90}
print(student_copy) # Output: {'name': 'Charlie', 'age': 26, 'score': 90}   
```

```python
## iterating dictionary
for key in student:
    print(key, student[key])
for key, value in student.items():
    print(key, value)
```

```python
## nested dictionary
students={
    "student1":{"name":"Alice", "age":25},
    "student2":{"name":"Bob", "age":26},
    "student3":{"name":"Charlie", "age":27}
}
print(students)
print(students["student1"]["name"]) # Output: Alice
```

```python
## dictionary comprehension
squares={x:x**2 for x in range(1,6)}
print(squares) # Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
## contoh comprehension dengan kondisi (hanya bilangan genap)
even_squares = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print(even_squares)  # Output: {2: 4, 4: 16, 6: 36, 8: 64, 10: 100}

## comprehension dari dictionary yang sudah ada (filter nilai numerik)
numeric_student_data = {k: v for k, v in student.items() if isinstance(v, int)}
print(numeric_student_data)  # Output: {'age': 26, 'score': 90}
```
