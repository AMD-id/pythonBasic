func merupakan block of code yang mengerjakan specific task. fungsi membantu dalam mengatur kode, menggunakan kembali kode, dan meningkatkan keteracaan kode

```python
## syntax
def function_name(parameters):
    """docstring"""
    # code block
    return value
```

```python
## kenapa fungsi penting?
# 1. Reusability: Fungsi memungkinkan kita untuk menulis kode sekali dan menggunakannya berkali-kali tanpa harus menulis ulang kode tersebut.
# 2. Modularitas: Fungsi membantu memecah program menjadi bagian-bagian yang lebih kecil dan lebih mudah dikelola, sehingga meningkatkan keterbacaan dan pemeliharaan kode.
# 3. Abstraksi: Fungsi memungkinkan kita untuk menyembunyikan detail implementasi dan fokus pada apa yang dilakukan fungsi tersebut, bukan bagaimana cara kerjanya.
# 4. Pengorganisasian: Fungsi membantu mengorganisir kode dengan cara yang logis, sehingga memudahkan pengembangan dan debugging.
## contoh fungsi sederhana
def greet(name):
    """Fungsi untuk menyapa seseorang"""
    return f"Hello, {name}!"
print(greet("Alice"))  # Output: Hello, Alice!
```

```python
def num_odd_even(n):
    """Fungsi untuk menghitung jumlah bilangan ganjil dan genap dari 1 hingga n"""
    if n%2 == 0:
        print(n, "adalah bilangan genap")
    else:
        print(n, "adalah bilangan ganjil")
```

```python
num = 10
num_odd_even(num)
```

```python
num2 = 5341
num_odd_even(num2)
```

```python
## fuction dengan beberapa parameter

def cal_number(num1, num2, num3):
    """Fungsi untuk menghitung jumlah tiga bilangan"""
    return num1 + num2 + num3
```

```python
print(cal_number(5, 10, 15))
```

```python
## default parameter

def greet(name="Guest"):
    """Fungsi untuk menyapa seseorang dengan nama default"""
    return f"Hello, {name}!"

print(greet())  # Output: Hello, Guest!
```

```python
## varible length argument
## return value dengan variable length argument
def sum_all(*args):
    """Fungsi untuk menghitung jumlah dari semua argumen yang diberikan"""
    return sum(args)

print(sum_all(1, 2, 3, 4, 5))  # Output: 15
```

```python
## keyword argument
def print_detail(**kwargs):
    """Fungsi untuk mencetak detail dengan keyword argument"""
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_detail(name="Alice", age=30, city="New York")
```

```python
print_detail(name="Bob", age=25, city="Los Angeles", profession="Engineer")
```

```python
## bagaimana jika *args dan **kwargs digunakan bersamaan dalam satu fungsi?
def combined_example(*args, **kwargs):
    """Fungsi untuk menunjukkan penggunaan *args dan **kwargs bersama-sama"""
    print("Positional arguments (args):", args)
    print("Keyword arguments (kwargs):", kwargs)

combined_example(1, 2, 3, name="Charlie", age=28)
```
