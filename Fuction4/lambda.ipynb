## Lambda Fucntion

lambda merupakan fungsi kecil anonimus yang menggunakan keyword lambda, dapat memiliki berbagai argyment tp hanya bisa 1 expression, biasanya digunakan untuk operasi yang simple ataupun argumen 

```python
# Syntax

lambda arguments: expression
```

```python
## cotoh fungsi normal

def addition(x, y):
    return x + y   
    
```

```python
addition(5, 3)
```

```python
## function lambda
addition=lambda a,b: a + b
type(addition)
addition(5, 3)
```

```python
def even(num):
    if num%2==0:
        return True
```

```python
even(4)
```

```python
even1=lambda num:num%2==0
even1(20)
```

```python
def addition_plus(x,y,z):
    return x+y+z
    
```

```python
addition_plus1 = lambda x,y,z:x+y+z
```

```python
addition_plus1(3,4,5)
```

```python
## map() 
## apply a fuction to all items in a list (or any iterable) and return a map object (which is an iterator)
num=[1,2,3,4,5]
def square(number):
    return number**2

square(2)
```

```python
## map() dengan lambda
map(lambda number:number**2,num) # mengembalikan map object
```

```python
list(map(lambda number:number**2,num)) # mengubah map object menjadi list
```

```python
## map multiple iterables
num1=[1,2,3]
num2=[4,5,6]

added_number=list(map(lambda x,y:x+y,num1,num2))
print(added_number)
```
